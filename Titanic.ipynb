# **Question:**


##### Which class had the most survivors?




## **Study Description:**


##### This will require a census study because it needs to gather data from the whole passenger population of the Titanic, and I have access to that entire population in my dataset.


###### Here is the data that needs to be collected from the titanic.csv file:
###### - The whole passenger population's class
###### - Whether they survived or not

```python
import pandas as pd
```

```python
titanic = pd.read_csv("titanic.csv")
```

```python
class_1 = titanic[titanic["Pclass"] == 1]
class1_survivors = class_1["Survived"].sum()
print(class1_survivors)
```

```python
class_2 = titanic[titanic["Pclass"] == 2]
class2_survivors = class_2["Survived"].sum()
print(class2_survivors)
```

```python
class_3 = titanic[titanic["Pclass"] == 3]
class3_survivors = class_3["Survived"].sum()
print(class3_survivors)
```

```python
print(max(class1_survivors, class2_survivors, class3_survivors))
```

# Conclusion
##### Since class 1 had 136 survivors, which was the most out of all three classes, class 1 had the most survivors on the *Titanic*
