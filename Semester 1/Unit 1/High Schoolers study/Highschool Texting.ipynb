# Conclusion

##### There is no association between texting and the type of internet connection for high schoolers in the US, except for those who have no internet connection. 
##### - This can be seen with the conditional relative frequencies within each internet type for texting as the preferred method of communication. The conditional frequency of texting within "Yes - broadband connection" was 46%; within "Yes - dial-up connection" was 44%; and within  "Yes - other" was 45%. All three of these frequencies were within 5% of each other, which means that there is no significant difference between the percentages. Only the relative conditional frequency of texting within "No Internet" was significantly different from the rest, which was 29% (which had a difference of more than 5% when compared to the other texting within internet type frequencies). 
##### - So we can say that although there is no association between the types of internet and texting within high schoolers in the US, there is an association between the availability of internet and texting for high schoolers. High schoolers in the US who have internet are more likely to use texting as their preferred method of communication, while those who do not have internet are significantly less likely to use texting as their preferred method of communication.

Both "No Internet Connection" and "Yes - Dial up connection" were ommited for ease of making the graph


#### We can also see this in the dual bar chart above. Between the two most popular types of internet, there are differences in the frequencies of the preferred communication methods. However, they are proportional. In each category, the frequency of "Yes - other" internet is approximately 60% of the frequency of "Yes - broadband connection" internet, so they both have a similar frequency of text messaging within their specific internet. So there is no association between the type of internet connection and the preferred communication method being Text messaging.
