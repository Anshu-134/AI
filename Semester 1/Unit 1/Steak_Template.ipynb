### My Question
***Do people who have a college degree go skydiving more than those without a college degree?***.

### My Study
This is an observational study because we look at a sample of people and whether those people have gone skydiving and whether they have a college degree or higher. We are not looking at a whole population.

### My Analysis

```python
import pandas as pd
```

```python
df = pd.read_csv("steak.csv")
```

```python
skydive_yes = df[df["Have you ever been skydiving?"] == "Yes"]
skydive_no = df[df["Have you ever been skydiving?"] == "No"]
```

```python
skydive_yes_college = skydive_yes["Education"].value_counts()["Some college or Associate degree"]
skydive_yes_college += skydive_yes["Education"].value_counts()["Bachelor degree"]
skydive_yes_college += skydive_yes["Education"].value_counts()["Graduate degree"]
```

```python
skydive_no_college = skydive_no["Education"].value_counts()["Some college or Associate degree"]
skydive_no_college += skydive_no["Education"].value_counts()["Bachelor degree"]
skydive_no_college += skydive_no["Education"].value_counts()["Graduate degree"]
```

```python
"Less than high school degree" in skydive_yes["Education"].values
```

```python
"High school degree" in skydive_yes["Education"].values
```

```python
skydive_yes_collegeNO = skydive_yes["Education"].value_counts()["High school degree"]
```

```python
skydive_no_collegeNO = skydive_no["Education"].value_counts()["High school degree"]
skydive_no_collegeNO += skydive_no["Education"].value_counts()["Less than high school degree"]
```

```python
print("People who have gone to college and skydived")
print(skydive_yes_college)
print("People who have gone to college and not skydived")
print(skydive_no_college)
print("People who have not gone to college but have skydived")
print(skydive_yes_collegeNO)
print("People who have not gone to college or skydived")
print(skydive_no_collegeNO)

print()
print("Percent of those skydived out of those who went to college:")
print(skydive_yes_college/(skydive_yes_college+skydive_no_college)*100)
print("Percent of those skydived out of those who didn't go to college:")
print(skydive_yes_collegeNO/(skydive_yes_collegeNO+skydive_no_collegeNO)*100)
```

### My Conclusion
1. Yes, more people who go to college go skydiving. About 7.05% of people with a college degree have gone skydiving, but only about 2.44% of those who don't have a college degree have gone skydiving.
