# **Question:**
##### Which Gill Size of mushrooms is more common in Enlarging Stalk Shape mushrooms and Tapering Stalk Shape mushrooms?



### **Type of study:**
##### This is an Observational Study. I will be looking at the relative frequencies of different Gill sizes of the two Stalk shapes of mushrooms in a sample of  8,124 mushrooms. These represent the population (instead of actually being the population) of mushrooms in the Agaricus and Lepiota families, which is why it is an observational study instead of a census study.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fungi = pd.read_csv("mushrooms.csv")
```

```python
enlarging_shape = fungi[fungi["stalk-shape"] == "e"]
proportion_gill_e = enlarging_shape["gill-size"].value_counts(normalize=True)
print(f"{round(proportion_gill_e.b * 100, 2)}% of enlarging shape mushrooms have broad gills")
print(f"{round(proportion_gill_e.n * 100, 2)}% of enlarging shape mushrooms have narrow gills")
```

```python
tapering_shape = fungi[fungi["stalk-shape"] == "t"]
proportion_gill_t = tapering_shape["gill-size"].value_counts(normalize=True)
print(f"{round(proportion_gill_t.b * 100, 2)}% of enlarging shape mushrooms have broad gills")
print(f"{round(proportion_gill_t.n * 100, 2)}% of enlarging shape mushrooms have narrow gills")
```

```python
labels = 'Broad Gills in Enlarging Stalk Shape Mushrooms', 'Narrow Gills in Enlarging Stalk Shape Mushrooms'
sizes = [proportion_gill_e.b, proportion_gill_e.n]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['olivedrab', 'rosybrown', 'gray', 'saddlebrown'])
```

```python
labels = 'Broad Gills in Tapering Stalk Shape Mushrooms', 'Narrow Gills in Tapering Stalk Shape Mushrooms'
sizes = [proportion_gill_t.b, proportion_gill_t.n]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%')
```

# Conclusion

##### - Both stalk shapes of the mushrooms (enlarging and tapering) had a higher percentage of broad gills (80% of enlarging and 60% of tapering) than narrow gills (20% of enlarging and 40% of tapering). This can also be seen on the pie charts, as the "Broad Gills" part covers more than half of the pie chart in both shapes.
##### - Therefore, we can say that Broad Gills is the most common Gill Size in both stalk shapes of mushrooms. Additionally, we can also say that Broad Gills are more common in Enlarging Stalk Shape mushrooms than in Tapering Stalk Shape mushrooms. This is because Enlarging Stalk Shape mushrooms have a higher percentage of mushrooms with Broad Gills (at 80%) than Tapering Stalk Shape mushrooms (which only has 60% which is less than 80%).
