# Question:
##### Does the stalk shape of a mushroom affect its gill size?



### Type of study:
##### Relative frequency study: This is because I need to find the proportion of different gill sizes within two different stalk shapes, not the number, as there are different numbers of each stalk shape. Just comparing the frequencies in each stalk shape category would lead to an incorrect analysis.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fungi = pd.read_csv("mushrooms.csv")
```

```python
enlarging_shape = fungi[fungi["stalk-shape"] == "e"]
proportion_gill_e = enlarging_shape["gill-size"].value_counts(normalize=True)
print(f"{round(proportion_gill_e.b * 100, 2)}% of enlarging shape mushrooms have broad gills")
print(f"{round(proportion_gill_e.n * 100, 2)}% of enlarging shape mushrooms have narrow gills")
```

```python
tapering_shape = fungi[fungi["stalk-shape"] == "t"]
proportion_gill_t = tapering_shape["gill-size"].value_counts(normalize=True)
print(f"{round(proportion_gill_t.b * 100, 2)}% of enlarging shape mushrooms have broad gills")
print(f"{round(proportion_gill_t.n * 100, 2)}% of enlarging shape mushrooms have narrow gills")
```

```python
labels = 'Broad Gills in Enlarging Stalk Shape Mushrooms', 'Narrow Gills in Enlarging Stalk Shape Mushrooms'
sizes = [proportion_gill_e.b, proportion_gill_e.n]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['olivedrab', 'rosybrown', 'gray', 'saddlebrown'])
```

```python
labels = 'Broad Gills in Tapering Stalk Shape Mushrooms', 'Narrow Gills in Tapering Stalk Shape Mushrooms'
sizes = [proportion_gill_t.b, proportion_gill_t.n]

fig, ax = plt.subplots()
ax.pie(sizes, labels=labels, autopct='%1.1f%%')
```

# Conclusion

##### - Both stalk shapes of the mushrooms (enlarging and tapering) had a higher percent of broad gills (80% of enlarging and 60% of tapering) than narrow gills (20% of enlarging and 40% of tapering). This suggests that the stalk shape of a mushroom does not affect the gill size of that mushroom.
##### - However, it can be said that a higher percent of enlarging-stalk-shaped mushrooms have broad gills than tapering-stalk-shaped mushrooms. We can also say that a higher percent of tapering-stalk-shaped mushrooms have narrow gills than enlarging-stalk-shaped mushrooms.
